```python
import math
```

```python
e = math.e
x = 3

e_power_minus_4 = e ** -4
```

```python
poission_pmf = (e_power_minus_4 * (4 ** x)) / math.factorial(x)

poission_pmf
```

```python
def sigma_poisson(n):
    total = 0.0
    for i in range(n + 1):
        total += (e_power_minus_4 * (4 ** i)) / math.factorial(i)
    return total
```

```python
print(1 - sigma_poisson(3))
```
