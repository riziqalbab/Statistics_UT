## SOAL 1

### PELUANG DISKRET

- Peluang diskret adalah peluang yang terjadi pada suatu kejadian ketika jumlah kejadian dapat dihitung atau terbatas

Kita ambil contoh yaitu menghitung kemungkinan siswa di kelas mendapatkan ranking 1, 2, atau 3. Misalnya, jika ada 30 siswa di kelas, maka peluang diskret untuk mendapatkan ranking 1 adalah 1/30, untuk ranking 2 adalah 1/30, dan untuk ranking 3 adalah 1/30. Peluang diskret ini dapat dihitung dengan menggunakan rumus peluang yaitu:

$$
P(A) = \frac{\text{Jumlah Kejadian yang Diinginkan}}{\text{Jumlah Kejadian yang Mungkin}}
$$



### PELUANG KONTINU

- Peluang kontinu adalah peluang yang terjadi pada suatu kejadian ketika jumlah kejadian tidak dapat dihitung atau tidak terbatas. Dalam artian lain, peluang kontinu terjadi pada kejadian yang memiliki jumlah hasil yang tak terbatas atau dapat dihitung dengan menggunakan fungsi densitas probabilitas.

Kita bisa meneruskan pada siswa di kelas. Namun kita akan menghitung peluang siswa mendapatkan tinggi badan tertentu. Misalnya, jika kita ingin menghitung peluang siswa memiliki tinggi badan antara 150 cm dan 160 cm, dengan rata rata tinggi badan siswa di kelas adalah 155 cm dan standar deviasi adalah 5 cm, maka kita dapat menggunakan distribusi normal untuk menghitung peluang kontinu tersebut. Kita dapat menggunakan rumus distribusi normal yaitu:


$$
f(x) = \frac{1}{\sigma \sqrt{2\pi}} e^{-\frac{1}{2} \left(\frac{x - \mu}{\sigma}\right)^2}
$$
Jawaban: 

- Melakukan tranformasi x menjadi z dengan rumus:
$$z = \frac{x - \mu}{\sigma}
$$
- Kemudian kita hitung z untuk batas bawah (150 cm) dan batas atas (160 cm):
$$z_1 = \frac{150 - 155}{5} = -1
$$
$$z_2 = \frac{160 - 155}{5} = 1
$$

$$P(150 < X < 160) = P(Z < 1) - P(Z < -1) = 0.8413 - 0.1587 = 0.6826
$$

## SOAL 3

- DIKETAHUI
    - Pabrik LED memproduksi 12.000 unit, diperkirakan 8 unit yang cacat. Toko emmesan 6.000 unit untuk dijual

- A. Toko Jaya menerima paling banyak 2 lampu cacat (Hitunglah dengan pendekatan distribusi binomial)
- B. Toko Jaya menerima lebih dari 3 lampu cacat (Hitunglah dengan pendekatan distribusi poisson)
- C. Tentukan rata-rata dan simpangan baku dari banyaknya lampi cacat yang diterima Toko Jaya sesuai dengan bagian A dan B


### JAWABAN

#### JAWABAN A
$$
P(X = k) = \binom{n}{k} p^k (1-p)^{n-k}
$$
Dimana:
- $n$ adalah jumlah percobaan (jumlah lampu yang diterima, yaitu
6000)
- $k$ adalah jumlah keberhasilan (jumlah lampu cacat yang diterima, yaitu0, 1, atau 2)
- $p$ adalah peluang keberhasilan (peluang lampu cacat, yaitu 8/12000)
- $1-p$ adalah peluang kegagalan (peluang lampu tidak cacat, yaitu $1 - 8/12000)$
- Untuk menghitung peluang Toko Jaya menerima paling banyak 2 lampu cacat, kita perlu menghitung peluang untuk k = 0, k = 1, dan k = 2, kemudian menjumlahkan hasilnya:
$$P(X \leq 2) = P(X = 0) + P(X = 1) + P(X = 2)
$$





#### JAWABAN B
$$ \mu = np $$
$$ \mu = 6000 \cdot \frac{8}{12000} = 4$$
$$P(X > 3) = 1 - P(X \leq 3) = 1 - \sum_{k=0}^{3} \frac{e^{-\mu} \mu^k}{k!}$$



- Selanjutnya, kita hitung peluang untuk $k = 0, 1, 2, 3$:

$$
P(X \leq 3) = \sum_{k=0}^{3} \frac{e^{-4} 4^k}{k!}
$$

$$
P(X \leq 3) = e^{-4}\left(1 + 4 + \frac{4^2}{2!} + \frac{4^3}{3!}\right)
$$

$$
P(X > 3) = 1 - P(X \leq 3)
$$

$$
P(X > 3) = 1 - e^{-4}\left(1 + 4 + \frac{4^2}{2!} + \frac{4^3}{3!}\right)
$$

$$
P(X > 3) \approx 0.5665
$$



- Jadi, peluang Toko Jaya menerima lebih dari 3 lampu cacat adalah **0.5665 atau 56.65%**.




#### JAWABAN C
$$\mu = n \cdot p
$$
- Simpangan baku (standar deviasi) dari distribusi binomial dapat dihitung dengan rumus:
$$\sigma = \sqrt{n \cdot p \cdot (1-p)}
$$
- Untuk bagian A, kita menggunakan nilai n = 6000 dan p = 8/12000 untuk menghitung rata-rata dan simpangan baku:
$$\mu_A = 6000 \cdot \frac{8}{12000} = 4
$$
$$\sigma_A = \sqrt{6000 \cdot \frac{8}{12000} \cdot \left(1 - \frac{8}{12000}\right)} \approx 1.999
$$
- Untuk bagian B, kita menggunakan nilai n = 6000 dan p = 8/12000 untuk menghitung rata-rata dan simpangan baku:
$$\mu_B = 6000 \cdot \frac{8}{12000} = 4
$$
$$\sigma_B = \sqrt{6000 \cdot \frac{8}{12000} \cdot \left(1 - \frac{8}{12000}\right)} \approx 1.999
$$
- Jadi, rata-rata dan simpangan baku dari banyaknya lampu cacat yang diterima Toko Jaya sesuai dengan bagian A dan B adalah:
    - Rata-rata (mean) = 4
    - Simpangan baku (standar deviasi) ≈ 1.999



## SOAL 4

DIKETAHUI
- n = 300
- p = 0.2
- q = 1 - p = 0.8

X = Jumlah email yang dibuka

Pertanyaan:
- Peluang bahwa jumalah email yang dibuka berada pada rentang antar 45 samoai 75 email termasuk 45 dan 75

### JAWABAN

$$ \mu = n \cdot p $$
$$ \mu = 300 \cdot 0.2 = 60 $$
$$ \sigma = \sqrt{300 \cdot 0.2 \cdot 0.8} = \sqrt{48} \approx 6.93 $$

Gunakan pendekatan normal dengan koreksi kontinuitas:

$$
P(45 \le X \le 75) \approx P(44.5 \le X \le 75.5)
$$

Ubah ke bentuk $Z$:

$$
Z_1 = \frac{44.5 - 60}{6.93} \approx -2.24
$$

$$
Z_2 = \frac{75.5 - 60}{6.93} \approx 2.24
$$

Sehingga:

$$
P(45 \le X \le 75) \approx P(-2.24 \le Z \le 2.24)
$$


$$
P(Z \le 2.24) \approx 0.9875
$$

$$
P(Z \le -2.24) \approx 0.0125
$$

Maka:

$$
P(-2.24 \le Z \le 2.24) = 0.9875 - 0.0125 = 0.9750
$$

Jadi, peluang jumlah email yang dibuka berada antara 45 sampai 75 adalah:

$$
0.9750 \approx 97.50\%
$$
