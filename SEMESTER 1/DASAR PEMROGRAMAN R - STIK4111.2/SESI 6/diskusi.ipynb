```r

library(ggplot2)

tiket_pesawat <- read.csv("tiket_pesawat.csv")

ggplot(data = tiket_pesawat, aes(x = reorder(Maskapai, Jumlah_Tiket), y = Jumlah_Tiket, fill = Maskapai)) +
  geom_bar(stat = "identity", alpha = 0.8) +
  labs(title = "Perbandingan Jumlah Tiket Per Maskapai",
       x = "Maskapai",
       y = "Jumlah Tiket") +
  theme_minimal()
```
