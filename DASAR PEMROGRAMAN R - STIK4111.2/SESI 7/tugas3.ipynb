## NOMOR 1

```r
x <- seq(-3, 3, length.out = 300)
density <- dnorm(x)

plot(
  x, density,
  type = "l",
  xlab = "x",
  ylab = "density",
)

abline(h = 0, col = "red", lwd = 2)
```
