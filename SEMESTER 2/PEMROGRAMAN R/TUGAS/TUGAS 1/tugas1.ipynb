## A. Menghitung luas permukaan kerucut dan volume kerucut.

LP = PI * r * (r+s) <br>
V = 1/3 * pi * r^2 * t

```r
luas_permukaan_kerucut <- function(r, s){
    hasil <- pi * r * (r+s)

    return(hasil)
}

volume_kerucut <- function(r, t){
    hasil <- (1/3) * pi * (r**2) * t

    return(hasil)
}
```

```r
luas_permukaan_kerucut(10, 5)
```

```r
volume_kerucut(40, 10)
```

## B. Menghitung volume silinder.

V = pi * r^2 * t

```r
volume_silinder <- function(r, t){
    hasil <- pi * (r**2) * t

    return(hasil)
}
```

```r
volume_silinder(70, 50)
```

## C. 

Menghitung volume silinder.
Mensimulasikan nilai y pada suatu persamaan eksponensial 𝑦 = a⋅ebx + 𝑐 dengan nilai input  𝑥 mulai dari -1 hingga 5 dengan kenaikan 0.5 unit. Diketahui nilai  𝑎 = 5,  𝑏 = 0.8, dan  𝑐 = 2. 


```r
a <- 5
b <- 0.8
c <- 2


x <- seq(from = -1, to = 5, by = 0.5)

y <- a * exp(b*x) + c

result <- data.frame(x=x, y=y)


plot(result, col='red', type = 'b')
```
