## Menghitung luas permukaan kerucut dan volume kerucut.

LP = PI * r * (r + s) <br>
V = PI * r^2 * t * 1/3

```r
luas_permukaan_kerucut <- function(r, s){
    hasil <- pi * r * (r+s)

    return(hasil)
}

volume_kerucut <- function(r, t){
    hasil <- (pi*(r**2) * t)/3

    return(hasil)
}
```

```r
hasil_luas_permukaan_kerucut <- luas_permukaan_kerucut(10, 5)
hasil_volume_kerucut <- volume_kerucut(15, 10)



print(hasil_luas_permukaan_kerucut)
print(hasil_volume_kerucut)
```

### Buatlah perintah di R untuk membuat plot fungsi  dengan dari rentang x = [0,6] Tambahkan garis vertikal pada nilai x = 3 untuk menunjukkan posisi titik tengah atau kandidat titik optimum fungsi.


```r
g <- function(x)(x**2) - 6*x + 8

curve(g, from=0, to=6, col='red')

abline(v=3, col='red')
```

### Buatlah perintah di R untuk menentukan dua akar persamaan dari fungsi g(x) = 0.


```r
akar <- polyroot(c(8, -6, 1))

akar_real <- Re(akar)

print(akar_real)
```

### Buatlah perintah di R untuk memperoleh nilai optimum fungsi g(x) pada selang x = [0, 6]. Tentukan pula nilai x yang menghasilkan nilai optimum tersebut.


```r
nilai <- optimize(g, interval = c(0, 6))

nilai$minimum
```

### Tampilkan hasil perhitungan nilai optimum dalam bentuk output yang memuat nilai x optimum dan nilai g(x) optimum.

```r
x_optimum <- nilai$minimum
gx_optimum <- nilai$objective

cat("Nilai x optimum : ", x_optimum, "\n")
cat("Nilai g(x) optimum:", gx_optimum)
```

```r
n <- 15
p <- 0.4
x <- seq(0, n)


pmf <- dbinom(x, size = n, prob = p)

plot(pmf, col='red', main="Grafik PMF Distribusi Binomial", xlab="Banyak Siswa Yang Menjawab Benar", ylab=" P(X = x)")
```

```r
cdf <- pbinom(x, size = n, prob = p)

plot(cdf, main="Grafik CDF Distribusi Binomial(n =15, p=0.4)", xlab="Banyak Siswa Yang Menjawab Benar", ylab="P(X <= x)")
```
