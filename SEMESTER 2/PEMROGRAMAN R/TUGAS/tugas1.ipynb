### A. Menghitung luas permukaan kerucut dan volume kerucut.

LP = pi * r * (r+s) <br>
V = 1/3 *pi * r^2 x t

```r
luas_permukaan <- function(r, s){
    hasil <- pi * r * (r + s)


    return(hasil)
}

volume_kerucut <- function(r, t){
    hasil <- (1/3) * pi * (r**2) * t

    return(hasil)
}
```

```r
luas_permukaan(12, 10)
volume_kerucut(10, 10)
```

### B. Menghitung volume silinder.


V = pi * r^2 * t

```r
volume_silinder <- function(r, t){
    hasil <- pi * (r**2) * t 

    return(hasil)
}
```

```r
volume_silinder(45, 10)
```
