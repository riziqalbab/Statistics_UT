```r
x <- 0:15

distribusi_a <- dbinom(x, size = 15, prob = 0.3)
distribusi_b <- dbinom(x, size = 15, prob = 0.6)

grafik <- barplot(distribusi_a, names.arg = x, col='blue', main = "Perbandingan Dua Distribusi Binomial", xlab = "Jumlah Keberhasilan", ylab = "Peluang")

lines(distribusi_b, col='red', type='b', pch=16)

legend("topright", legend = c("Distribusi A: B(15, 0.3)", "Distribusi A: B(15, 0.6)"), col = c("blue", "red"), pch = c(15, 16), lty = c(NA, 1))
```

## SOAL 2

```r
tahun <- 2013:2022
inflasi <- c(5.2, 4.8, 6.5, 7.1, 5.9, 4.7, 6.2, 6.8, 5.5, 4.9)

nilai_intervensi <- 6

index_tahun_6_persen <- which(inflasi > 6)[1]


grafik_data_inflasi <- barplot(inflasi, names.arg = tahun, ylab = "Inflasi (%)", xlab = "Tahun", main="Grafik Inflasi Selama 10 Tahun Terakhir")

abline(h= nilai_intervensi, col='red', lty=2)
abline(v=grafik_data_inflasi[index_tahun_6_persen], lty=2, col="darkgreen")

points(grafik_data_inflasi[index_tahun_6_persen], inflasi[index_tahun_6_persen], col='red', pch=19, cex=1.5)

legend('topright', legend = c("Batas Intervensi (6%)", "Tahun Pertama Melebihi Batas", "Titik Inflasi Pertama > 6%"), col=c("red", "darkgreen", 'red'), lty=c(2,2,NA), pch=c(NA, NA, 19), bty = 'n')
```

## SOAL 3

```r
A <- matrix(c(1, 0, -3, -1, 4, 8, 0, 0, 3, 0, 6, 0, 9, 5, 7, -2), nrow = 4, byrow = TRUE)

A
```

```r
trace_A <- sum(diag(A))
trace_A
```

```r
determinan_A <- det(A)


if(determinan_A != 0 ){
    print("Matrix ini adalah non singular")
} else{
    print("Matrix ini adalah singular")
}
```

```r
invers_A <- solve(A)

invers_A
```

```r
eigen_A <- eigen(A)

eigen_A$values
```
